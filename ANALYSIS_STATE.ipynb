## 3. Rencana Pemisahan Fitur (anti-sirkularitas)

Fitur clustering dan fitur klasifikasi **dipisah** supaya model klasifikasi tidak sekadar menghafal label yang dibuat dari fitur yang sama.

| Blok | Dipakai untuk | Contoh fitur |
|---|---|---|
| **A. Fitur risiko/outcome** | Clustering (membentuk label) | incidence rate ISPA per 1.000 penduduk, rate pneumonia balita, proporsi kasus berat |
| **B. Fitur kontekstual** | Klasifikasi (prediktor) | kepadatan penduduk, sanitasi layak, air minum layak, rasio fasilitas kesehatan, rasio tenaga kesehatan, IPM, garis kemiskinan |

**Aturan:** tidak ada kolom yang masuk ke kedua blok. Kalau jumlah fitur kontekstual terbatas (< 4), blok B boleh ditambah variabel yang tidak dipakai saat clustering, tetapi tetap tidak boleh memuat outcome.

**Catatan n = 38:** jumlah fitur clustering dibatasi **3–5** agar jarak antar-observasi tetap bermakna (hindari curse of dimensionality), dan standardisasi z-score wajib sebelum K-Means.

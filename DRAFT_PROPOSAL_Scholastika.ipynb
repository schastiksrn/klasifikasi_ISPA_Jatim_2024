### F.1.3 Deskripsi Variabel

Variabel dikelompokkan menjadi dua blok. **Blok A** dipakai untuk membentuk label melalui clustering, sedangkan **Blok B** dipakai sebagai prediktor pada tahap klasifikasi. Pemisahan ini dilakukan agar label tidak diprediksi dari variabel yang sama dengan pembentuk labelnya.

| Blok | Variabel | Definisi Operasional | Satuan | Tipe | Sumber |
|---|---|---|---|---|---|
| A | `rate_pneu` | Jumlah penemuan kasus pneumonia pada balita (laki-laki + perempuan) dibagi jumlah balita, dikali 1.000 | kasus per 1.000 balita | Rasio | Profil Kesehatan |
| A | `prop_berat_pct` | Jumlah kasus pneumonia berat pada balita dibagi total penemuan pneumonia balita, dikali 100 | persen | Rasio | Profil Kesehatan |
| B | `iku` | Indeks Kualitas Udara, indeks komposit kualitas udara ambien | poin 0–100 | Rasio | Satu Data Jatim |
| B | `sanitasi` | Persentase rumah tangga dengan akses sanitasi layak | persen | Rasio | Profil Kesehatan |
| B | `kepadatan` | Jumlah penduduk dibagi luas wilayah | jiwa/km² | Rasio | BPS |

Rumus dua variabel pembentuk label:

$$\text{rate\_pneu}_i = \frac{\text{penemuan pneumonia balita}_i}{\text{jumlah balita}_i} \times 1{.}000$$

$$\text{prop\_berat}_i = \frac{\text{pneumonia berat}_i}{\text{penemuan pneumonia balita}_i} \times 100$$

**Alasan pemilihan.**

1. **Dua variabel pada Blok A, bukan satu.** Bila label hanya dibentuk dari satu variabel, clustering menjadi satu dimensi dan hasilnya sekadar pemotongan urutan angka — bukan pengelompokan multidimensi.
2. **Rate, bukan jumlah kasus.** Jumlah kasus mentah sangat ditentukan oleh besar populasi, sehingga clustering berisiko hanya memisahkan wilayah besar dan kecil.
3. **Pneumonia balita sebagai indikator.** Pneumonia merupakan bentuk ISPA paling berat dan satu-satunya indikator ISPA yang tersedia hingga level kabupaten/kota pada publikasi resmi.
4. **Kepadatan, bukan jumlah penduduk.** Jumlah penduduk akan mengelompokkan wilayah berdasarkan besar-kecilnya, bukan berdasarkan tingkat kepadatan hunian yang relevan bagi penularan.
5. **Arah IKU perlu diperhatikan.** Nilai IKU yang tinggi menunjukkan udara yang lebih bersih, sehingga arah hubungannya berlawanan dengan variabel risiko lainnya. Hal ini diperhitungkan saat interpretasi, bukan dengan membalik nilainya, agar satuan tetap sesuai definisi resmi.

### F.2.5 Penanganan Pencilan

Pencilan dideteksi dengan aturan IQR:

$$\text{Batas bawah} = Q_1 - 1{,}5 \times IQR, \qquad \text{Batas atas} = Q_3 + 1{,}5 \times IQR$$

dengan $IQR = Q_3 - Q_1$.

| Variabel | Jumlah pencilan | Keterangan |
|---|---|---|
| `rate_pneu` | 2 | Termasuk Kota Blitar dengan 104,46 kasus per 1.000 balita |
| `prop_berat_pct` | 2 | — |
| `iku` | 1 | — |
| `sanitasi` | 2 | — |
| `kepadatan` | 9 | Terkonsentrasi pada kota-kota besar, dengan Surabaya 8.698 jiwa/km² |

Keputusan: **pencilan tidak dihapus**, dengan alasan sebagai berikut.

1. Pada konteks risiko penyakit, nilai ekstrem justru merupakan informasi utama — kabupaten/kota dengan beban pneumonia sangat tinggi, atau dengan kepadatan hunian sangat tinggi, adalah objek yang ingin diidentifikasi, bukan gangguan.
2. Menghapus observasi akan mengurangi n dari 38 menjadi lebih kecil, sehingga clustering menjadi tidak stabil.
3. Nilai ekstrem pada data ini bukan kesalahan pencatatan. Kota Blitar, misalnya, mencatat 1.090 kasus penemuan terhadap perkiraan 464 kasus, sehingga cakupan penemuannya 234,73 persen. Angka ini benar sesuai sumber, dan justru menunjukkan perbedaan kinerja penemuan kasus antarwilayah.
4. Pencilan ditangani melalui transformasi (F.2.6), bukan melalui penghapusan.

### F.2.6 Transformasi Data

**a. Pembentukan rate dan proporsi.** Jumlah kasus mentah tidak dapat dibandingkan langsung antarwilayah karena sangat dipengaruhi jumlah penduduk dan jumlah balita. Oleh karena itu dibentuk laju:

$$IR_i = \frac{K_i}{P_i} \times 1{.}000$$

dengan $K_i$ = jumlah penemuan kasus pneumonia balita di wilayah $i$ dan $P_i$ = jumlah balita di wilayah $i$. Tanpa langkah ini, clustering berisiko hanya memisahkan wilayah berdasarkan ukuran populasi.

Proporsi kasus berat dibentuk dengan cara serupa:

$$PB_i = \frac{B_i}{K_i} \times 100$$

dengan $B_i$ = jumlah kasus pneumonia berat di wilayah $i$.

**b. Transformasi logaritma.** Transformasi $x' = \log(1 + x)$ diterapkan pada variabel dengan kemencengan positif lebih dari 1, karena jarak Euclidean sangat peka terhadap nilai ekstrem pada sebaran seperti itu.

| Variabel | Skewness | Transformasi |
|---|---|---|
| `rate_pneu` | +1,82 | $\log(1+x)$ |
| `kepadatan` | +1,75 | $\log(1+x)$ |
| `prop_berat_pct` | −1,09 | tidak (kemencengan negatif) |
| `iku` | −1,14 | tidak (kemencengan negatif) |
| `sanitasi` | −0,87 | tidak |

Transformasi logaritma tidak diterapkan pada variabel yang menceng ke kiri, karena logaritma justru memperlebar jarak pada ekor kiri dan tidak memperbaiki kemencengan negatif. Variabel seperti itu ditangani melalui standardisasi pada F.2.8.

### F.2.7 Pengkodean

| Keperluan | Metode | Alasan |
|---|---|---|
| Identitas wilayah | Kode dan nama kabupaten/kota dipertahankan sebagai indeks, bukan fitur | Nama bukan besaran dan tidak boleh ikut dihitung jaraknya |
| Label cluster | Pengkodean ordinal: rendah = 0, sedang = 1, tinggi = 2 | Urutan kategori bermakna, sehingga ordinal lebih tepat daripada one-hot |
| Variabel kategorik (bila ada) | One-hot encoding | Menghindari kesan urutan pada kategori yang setara |

### F.2.8 Normalisasi

Sebelum clustering, seluruh fitur distandarkan dengan *z-score*:

$$z_{ij} = \frac{x_{ij} - \bar{x}_j}{s_j}$$

dengan $\bar{x}_j$ dan $s_j$ berturut-turut rata-rata dan simpangan baku variabel $j$.

Alasan: K-Means dan Ward sama-sama menggunakan jarak Euclidean. Tanpa standardisasi, variabel bersatuan besar (misalnya kepadatan penduduk dalam jiwa/km²) akan mendominasi variabel bersatuan kecil (misalnya persentase sanitasi), sehingga cluster yang terbentuk tidak lagi mencerminkan pola bersama antarfitur. Standardisasi juga membuat pencilan tidak lagi mendominasi jarak, sehingga keputusan tidak menghapus pencilan pada F.2.5 tetap dapat dipertanggungjawabkan.

Parameter standardisasi (rata-rata dan simpangan baku) dihitung **hanya dari data latih** pada tahap klasifikasi, agar tidak terjadi kebocoran informasi (*data leakage*) dari data uji.
